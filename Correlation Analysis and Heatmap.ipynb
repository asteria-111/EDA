# 10 & 11. Correlation Analysis & Visual Heatmap

In this section, we isolate all numerical variables to evaluate their underlying linear relationships. By computing a correlation matrix and plotting it as a visual heatmap, we can easily scan for strong dependencies (such as the connection between bounce and exit rates) and identify which variables have the most baseline influence on shopper outcomes.


```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the online shoppers behavior dataset
df = pd.read_csv('online_shoppers_intention.csv')

# Double check that the data loaded by showing the first 3 rows
df.head(3)
```

```python
numerical_df = df.select_dtypes(include=['int64', 'float64'])
correlation_matrix = numerical_df.corr()
plt.figure(figsize=(12, 10))

# We use 'coolwarm' color scheme: red is positive correlation, blue is negative correlation
sns.heatmap(
    correlation_matrix, 
    annot=True,             # Prints the exact correlation numbers inside the squares
    fmt=".2f",              # Rounds the numbers to 2 decimal places
    cmap='coolwarm',        # Classic red/blue e-commerce palette
    square=True,            # Forces squares to be perfectly equal
    linewidths=0.5,         # Adds a subtle white border line between squares
    cbar_kws={"shrink": .8} # Slightly shrinks the color bar slider legend on the side
)

plt.title('Correlation Heatmap of Online Shopper Behavioral Variables', fontsize=14, pad=15)
plt.tight_layout()
plt.show()
```
