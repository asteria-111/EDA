# 12. Final Visualization: Behavioral Funnel Drivers

To conclude our visual exploration, we build a multi-variable faceted plot. This final visualization maps **Page Values** against **Exit Rates** while splitting the data by **Visitor Type** and **Revenue**. It provides the ultimate birds-eye view of where shopper paths split into successful purchases versus abandoned carts.


```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the online shoppers behavior dataset
df = pd.read_csv('online_shoppers_intention.csv')

# Double check that the data loaded by showing the first 3 rows
df.head(3)
```

```python
sns.set_theme(style="ticks")
# Create a FacetGrid to plot side-by-side behavioral charts for New vs Returning Visitors
g = sns.FacetGrid(df, col="VisitorType", hue="Revenue", palette="Set1", height=5, aspect=1.2, col_order=['Returning_Visitor', 'New_Visitor'])

# Map a scatter plot into each grid quadrant comparing Page Values vs Exit Rates
g.map(sns.scatterplot, "ExitRates", "PageValues", alpha=0.7, edgecolor="w", linewidth=0.5)

# Add a unified legend box on the side
g.add_legend(title="Completed Purchase?")

# Clean up titles and labels dynamically
g.set_axis_labels("Page Exit Rate (%)", "Average Page Value ($)")
g.set_titles(col_template="Visitor Profile: {col_name}")

plt.subplots_adjust(top=0.8)
g.fig.suptitle('How Page Value and Exit Rates Dictate Final Sales across Shopper Segments', fontsize=14, y=0.95)

plt.show()
```
