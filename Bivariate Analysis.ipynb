# 8. Bivariate Analysis

In this section, we analyze the relationships between two variables. Specifically, we look at how numerical web metrics (like Page Values) and temporal factors (like Weekends) correlate with our target variable, **Revenue**, to uncover drivers behind successful transactions.


```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the online shoppers intention dataset
df = pd.read_csv('online_shoppers_intention.csv')

# Double check that the data loaded by showing the first 3 rows
df.head(5)
```

```python

# Set the theme style
sns.set_theme(style="whitegrid")

# 1. Plot Relationship between PageValues and Revenue
plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
sns.boxplot(data=df, x='Revenue', y='PageValues', hue='Revenue', palette='Set2', legend=False)
plt.title('Page Values vs. Revenue Outcome')
plt.xlabel('Purchased (Revenue)')
plt.ylabel('Page Value')

# 2. Plot Relationship between Weekends and Revenue (Following the new Seaborn warning rules!)
plt.subplot(1, 2, 2)
sns.countplot(data=df, x='Weekend', hue='Revenue', palette='autumn')
plt.title('Purchases Made on Weekends vs. Weekdays')
plt.xlabel('Is Weekend?')
plt.ylabel('Number of Sessions')

plt.tight_layout()
plt.show()
```

## 8.2 Additional Bivariate Relationship Analysis

To dive deeper into session dynamics, we evaluate:
1. **Bounce Rates vs. Revenue:** Investigating if high landing-page drop-off rates directly restrict financial conversions.
2. **Seasonal Shopping Streams (Month vs. Revenue):** Tracking if specific months display a higher historical conversion velocity compared to standard browsing cycles.


```python
plt.figure(figsize=(14, 5))

# Plot 1: Bounce Rates vs Revenue (Box Plot)
plt.subplot(1, 2, 1)
sns.boxplot(data=df, x='Revenue', y='BounceRates', hue='Revenue', palette='coolwarm', legend=False)
plt.title('Bounce Rates Breakdown by Conversion Outcome')
plt.xlabel('Purchased (Revenue)')
plt.ylabel('Bounce Rate Percentage')

# Plot 2: Monthly Conversion Counts (Following updated Seaborn warning rules!)
plt.subplot(1, 2, 2)
# Sorting months chronologically if needed, or letting categorical ordering apply
sns.countplot(data=df, x='Month', hue='Revenue', palette='Set1', order=['Feb', 'Mar', 'May', 'June', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec'])
plt.title('Monthly Distribution of Sessions and Sales')
plt.xlabel('Month of Session')
plt.ylabel('Total Sessions Count')

plt.tight_layout()
plt.show()
```
