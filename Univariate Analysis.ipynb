# 7. Univariate Analysis

In this section, we examine the independent distributions of single columns in our online shopper behavior dataset. This helps us understand the composition of our traffic and how long visitors stay on the site before looking at relationships between variables.


```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the online shoppers intention dataset
df = pd.read_csv('online_shoppers_intention.csv')

# Double check that the data loaded by showing the first 3 rows
df.head(3)
```

```python
# Create a bar chart following the updated Seaborn specifications
plt.figure(figsize=(7, 4))
sns.countplot(data=df, x='VisitorType', hue='VisitorType', palette='viridis', legend=False)
plt.title('Distribution of Site Visitor Profiles')
plt.xlabel('Visitor Category')
plt.ylabel('Number of Sessions')
plt.show()
```

```python
# 1. Get quick stats on time spent on product pages (Average, Min, Max)
print("--- Product Related Duration Stats ---")
print(df['ProductRelated_Duration'].describe())

# 2. Plot a histogram to see the distribution curve
plt.figure(figsize=(8, 4))
sns.histplot(data=df, x='ProductRelated_Duration', bins=50, kde=True, color='teal')
plt.xlim(0, 5000) # Zooms in so long tail values don't crush the chart layout
plt.title('Distribution of Time Spent on Product Related Pages')
plt.xlabel('Duration (Seconds)')
plt.ylabel('Number of Sessions')
plt.show()
```
