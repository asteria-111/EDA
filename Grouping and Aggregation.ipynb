# 9. Grouping and Aggregation

In this section, we transition from individual visual plots to definitive mathematical groupings. By aggregating behavioral variables across dimensions like **Month** and **VisitorType**, we can extract exact conversion baselines, average drop-off metrics, and page value intensities to guide commercial decision-making.


```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the online shoppers behavior dataset
df = pd.read_csv('online_shoppers_intention.csv')

# Double check that the data loaded by showing the first 3 rows
df.head(3)
```

```python
# --- Aggregation 1: Behavior Metrics by Month ---
print("--- Aggregated Shopping Metrics by Month ---")
monthly_summary = df.groupby('Month').agg({
    'Revenue': ['count', 'mean'],     # Total sessions and conversion rate
    'PageValues': 'mean',             # Average page importance value
    'BounceRates': 'mean'             # Average drop-off rate
}).reset_index()

# Flatten the multi-level columns for cleaner reading
monthly_summary.columns = ['Month', 'Total_Sessions', 'Conversion_Rate', 'Avg_Page_Value', 'Avg_Bounce_Rate']
# Format Conversion Rate to a readable percentage percentage
monthly_summary['Conversion_Rate'] = (monthly_summary['Conversion_Rate'] * 100).round(2)
monthly_summary[['Avg_Page_Value', 'Avg_Bounce_Rate']] = monthly_summary[['Avg_Page_Value', 'Avg_Bounce_Rate']].round(4)

print(monthly_summary.to_string(index=False))


# --- Aggregation 2: Purchase Behavior by Visitor Profile ---
print("\n--- Aggregated Metrics by Visitor Category ---")
visitor_summary = df.groupby('VisitorType').agg({
    'Revenue': 'mean',
    'PageValues': 'mean',
    'ExitRates': 'mean'
}).reset_index()

visitor_summary.columns = ['Visitor_Type', 'Conversion_Rate_%', 'Avg_Page_Value', 'Avg_Exit_Rate']
visitor_summary['Conversion_Rate_%'] = (visitor_summary['Conversion_Rate_%'] * 100).round(2)
visitor_summary[['Avg_Page_Value', 'Avg_Exit_Rate']] = visitor_summary[['Avg_Page_Value', 'Avg_Exit_Rate']].round(4)

print(visitor_summary.to_string(index=False))
```
